# 🎬 Torrentio Google Drive Yüksek Hızlı İndirici (Debrid Ingest)
Bu araç, 80 GB'a varan 4K Remux veya büyük torrentleri **Google'ın kendi yüksek hızlı veri merkezi (1 Gbps+)** üzerinden doğrudan 5 TB Google Drive hesabınızdaki `/Stremio` klasörüne aktarır.

İndirme tamamlandığında film, **Nuvio** ve **Stremio** eklentinizde en tepede **`[GDrive+]`** etiketiyle görünür ve Google CDN üzerinden **sıfır donma, sıfır P2P ve 1 Gbps hızla** oynatılır.

```python
# @title 1. Adım: Google Drive'ı Bağla 📁
# @markdown Sol taraftaki **Play (Çalıştır)** butonuna basın ve Google hesabınıza izin verin:

from google.colab import drive
import os

drive.mount('/content/drive')
stremio_dir = '/content/drive/MyDrive/Stremio'
os.makedirs(stremio_dir, exist_ok=True)

print("\n" + "="*50)
print("✅ Google Drive başarıyla bağlandı!")
print(f"📁 İndirme Klasörü: {stremio_dir}")
print("="*50)
```

```python
# @title 2. Adım: Magnet Linkini Yapıştır ve İndir 🚀
# @markdown İstediğiniz filmin/dizinin Magnet linkini buraya yapıştırıp sol taraftaki **Play (Çalıştır)** butonuna basın:

MAGNET_LINK = ""  # @param {type:"string"}

import os
import subprocess
import sys

if not MAGNET_LINK or not MAGNET_LINK.strip().startswith("magnet:"):
    print("❌ Lütfen geçerli bir Magnet linki yapıştırın! (magnet:?xt=... formatında olmalı)")
else:
    stremio_dir = '/content/drive/MyDrive/Stremio'
    os.makedirs(stremio_dir, exist_ok=True)

    if not os.path.exists('/usr/bin/aria2c'):
        print("⏳ Yüksek hızlı indirme motoru (Aria2) kuruluyor...")
        subprocess.run(['apt-get', 'update', '-qq'], stdout=subprocess.DEVNULL)
        subprocess.run(['apt-get', 'install', '-y', '-qq', 'aria2'], stdout=subprocess.DEVNULL)
        print("✅ Motor hazır!")

    print(f"\n⚡ Buluttan buluta indirme başlatılıyor...")
    print(f"📁 Hedef: Google Drive / Stremio")
    print("=" * 60)

    trackers = [
        'udp://tracker.opentrackr.org:1337/announce',
        'udp://open.stealth.si:80/announce',
        'udp://tracker.torrent.eu.org:451/announce',
        'udp://open.demonii.com:1337/announce',
        'udp://exodus.desync.com:6969/announce',
        'udp://tracker.skynetcloud.site:6969/announce',
        'udp://tracker.qu.ax:6969/announce'
    ]
    tracker_args = ['--bt-tracker=' + ','.join(trackers)]

    cmd = [
        'aria2c',
        '--enable-dht=true',
        '--dht-listen-port=6881',
        '--enable-peer-exchange=true',
        '--bt-enable-lpd=true',
        '--bt-max-peers=120',
        '--bt-tracker-connect-timeout=5',
        '--bt-tracker-timeout=5',
        '--max-connection-per-server=16',
        '--split=16',
        '--min-split-size=1M',
        '--file-allocation=none',
        '--summary-interval=5',
        '--seed-time=0',
        '--dir=' + stremio_dir
    ] + tracker_args + [MAGNET_LINK.strip()]

    subprocess.run(cmd)

    print("\n" + "=" * 60)
    print("🎉 TEBRİKLER! Dosya doğrudan Google Drive'ınıza indirildi.")
    print("🎬 Şimdi Nuvio veya Stremio'yu açın; filminiz en tepede [GDrive+] etiketiyle hazır!")
    print("=" * 60)
```
